```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../../../data/cleaned_data/cleaned_pull_count.csv').drop("Unnamed: 0", axis=1)
```

```python
pd.set_option('display.max_rows', None)

pd.set_option('display.max_columns', None)

pd.set_option('display.width', None)

pd.set_option('display.max_colwidth', None)
```

```python
df.shape
```

```python
df_encoded = pd.get_dummies(df, columns=['3D-model', 'rarity', 'weapon', 'element', 'region', 'asc_stat_bonus_category'], prefix=['model', 'rarity', 'weapon', 'element', 'region', 'asc_bonus'], dtype=int)
```

```python
df_encoded.shape
```

```python
df_encoded.drop(['rarity_4', 'weapon_Sword', 'element_Pyro', 'region_Extra-Terresterial', 'asc_bonus_Pyro DMG Bonus', 'model_Tall Male'], axis = 1, inplace=True)
df_encoded = df_encoded.rename(columns={'pulled_count_TARGET_VARIABLE': 'pulled_count', 'rarity_5': 'rarity_5_TARGET_VARIABLE'})
cols = ['rarity_5_TARGET_VARIABLE'] + [col for col in df_encoded.columns if col != 'rarity_5_TARGET_VARIABLE']
df_encoded = df_encoded[cols]
df_encoded['ones'] = 1
```

```python
df_encoded['pulled_count'] = np.log10(df_encoded['pulled_count']+1)

columns_to_normalize = ['pulled_count', 'lvl_90_HP', 'lvl_90_ATK', 'lvl_90_DEF', 'months_since_release', 'num_banners', 'avg_copies_per_player']
for col in columns_to_normalize:
    df_encoded[col] = (df_encoded[col] - df_encoded[col].min()) / (df_encoded[col].max() - df_encoded[col].min())
```

```python
df_encoded
```

```python
df_target = df_encoded['rarity_5_TARGET_VARIABLE']
df_inputs = df_encoded.drop('rarity_5_TARGET_VARIABLE', axis=1)
```

```python
print(df_target.shape)
print(df_inputs.shape)
```

```python
print(f"Class 0 count: {(df_target == 0).sum()}")
print(f"Class 1 count: {(df_target == 1).sum()}")
print(f"Class 1 percentage: {(df_target == 1).mean():.1%}")
```

```python
np.random.seed(42)

n_samples = len(df_inputs)
test_size = 0.3

indices = np.arange(n_samples)
np.random.shuffle(indices)

test_count = int(n_samples * test_size)
test_indices = indices[:test_count]
train_indices = indices[test_count:]

print(f'Total samples: {n_samples}')
print(f'Training samples: {len(train_indices)}')
print(f'Test samples: {len(test_indices)}')

df_inputs_train = df_inputs.iloc[train_indices].reset_index(drop=True)
df_inputs_test = df_inputs.iloc[test_indices].reset_index(drop=True)
df_target_train = df_target.iloc[train_indices].reset_index(drop=True)
df_target_test = df_target.iloc[test_indices].reset_index(drop=True)

print(f'\nTraining set - Class 0: {(df_target_train==0).sum()}, Class 1: {(df_target_train==1).sum()}')
print(f'Test set - Class 0: {(df_target_test==0).sum()}, Class 1: {(df_target_test==1).sum()}')
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500)))
```

```python
epsilon = 1e-7
```

```python
np.random.seed(42)
values = np.random.randn(2, 54) * 0.01
biases = np.zeros(2)
result_train = pd.DataFrame({
    'Linear1': (df_inputs_train * values[0]).sum(axis=1) + biases[0],
    'Linear2': (df_inputs_train * values[1]).sum(axis=1) + biases[1]
})
result_train['ReLU1'] = result_train['Linear1'].apply(lambda x: max(0, x))
result_train['ReLU2'] = result_train['Linear2'].apply(lambda x: max(0, x))
result_train['Logits'] = result_train['ReLU1'] + result_train['ReLU2']
result_train['Preds'] = sigmoid(result_train['Logits'])
result_train['Actual'] = df_target_train
result_train['Loss'] = -(df_target_train * np.log(result_train['Preds'] + epsilon) + (1 - df_target_train) * np.log(1 - result_train['Preds'] + epsilon))
loss_last = result_train['Loss'].mean()

print(f'Initial loss: {loss_last}')
```

```python
result_train
```

```python
max_iterations = 3000
learning_rate = 0.01

for counter in range(max_iterations):
    grad_w1 = np.zeros(54)
    grad_w2 = np.zeros(54)
    grad_b1 = 0
    grad_b2 = 0
    
    for row in range(len(df_inputs_train)):
        pred = result_train.loc[row, 'Preds']
        target = df_target_train.iloc[row]
        
        d_loss_d_logits = pred - target
        
        d_logits_d_relu1 = 1
        d_logits_d_relu2 = 1
        
        d_relu1_d_linear1 = 1 if result_train.loc[row, 'Linear1'] > 0 else 0
        d_relu2_d_linear2 = 1 if result_train.loc[row, 'Linear2'] > 0 else 0
        
        grad_from_relu1 = d_loss_d_logits * d_logits_d_relu1 * d_relu1_d_linear1
        grad_w1 += grad_from_relu1 * df_inputs_train.iloc[row].values
        grad_b1 += grad_from_relu1
        
        grad_from_relu2 = d_loss_d_logits * d_logits_d_relu2 * d_relu2_d_linear2
        grad_w2 += grad_from_relu2 * df_inputs_train.iloc[row].values
        grad_b2 += grad_from_relu2
    
    grad_w1 /= len(df_inputs_train)
    grad_w2 /= len(df_inputs_train)
    grad_b1 /= len(df_inputs_train)
    grad_b2 /= len(df_inputs_train)
    
    values[0] -= learning_rate * grad_w1
    values[1] -= learning_rate * grad_w2
    biases[0] -= learning_rate * grad_b1
    biases[1] -= learning_rate * grad_b2
    
    result_train['Linear1'] = (df_inputs_train * values[0]).sum(axis=1) + biases[0]
    result_train['Linear2'] = (df_inputs_train * values[1]).sum(axis=1) + biases[1]
    result_train['ReLU1'] = result_train['Linear1'].apply(lambda x: max(0, x))
    result_train['ReLU2'] = result_train['Linear2'].apply(lambda x: max(0, x))
    result_train['Logits'] = result_train['ReLU1'] + result_train['ReLU2']
    result_train['Preds'] = sigmoid(result_train['Logits'])
    result_train['Actual'] = df_target_train
    result_train['Loss'] = -(df_target_train * np.log(result_train['Preds'] + epsilon) + (1 - df_target_train) * np.log(1 - result_train['Preds'] + epsilon))
    loss_new = result_train['Loss'].mean()
    
    if counter % 10 == 0:
        print(f'Iteration: {counter}, Loss: {loss_new:.6f}')
    
    if abs(loss_new - loss_last) < 1e-7:
        print(f'Converged at iteration {counter}')
        break
    
    loss_last = loss_new

print(f'Final loss: {loss_new}')
```

```python
result_test = pd.DataFrame(index=range(len(df_inputs_test)))

result_test['Linear1'] = (df_inputs_test * values[0]).sum(axis=1) + biases[0]
result_test['Linear2'] = (df_inputs_test * values[1]).sum(axis=1) + biases[1]
result_test['ReLU1'] = result_test['Linear1'].apply(lambda x: max(0, x))
result_test['ReLU2'] = result_test['Linear2'].apply(lambda x: max(0, x))
result_test['Logits'] = result_test['ReLU1'] + result_test['ReLU2']
result_test['Preds'] = sigmoid(result_test['Logits'])

threshold = 0.65

train_predictions = (result_train['Preds'] >= threshold).astype(int)
test_predictions = (result_test['Preds'] >= threshold).astype(int)

train_accuracy = (train_predictions == df_target_train).mean()
test_accuracy = (test_predictions == df_target_test).mean()

print(f'\n=== RESULTS ===')
print(f'Training Accuracy: {train_accuracy:.2%}')
print(f'Test Accuracy: {test_accuracy:.2%}')

print(f'\nTest set prediction distribution:')
print(f'Class 0 predictions: min={result_test[df_target_test==0]["Preds"].min():.3f}, max={result_test[df_target_test==0]["Preds"].max():.3f}')
print(f'Class 1 predictions: min={result_test[df_target_test==1]["Preds"].min():.3f}, max={result_test[df_target_test==1]["Preds"].max():.3f}')

test_tp = ((test_predictions == 1) & (df_target_test == 1)).sum()
test_tn = ((test_predictions == 0) & (df_target_test == 0)).sum()
test_fp = ((test_predictions == 1) & (df_target_test == 0)).sum()
test_fn = ((test_predictions == 0) & (df_target_test == 1)).sum()

print(f'\nTest Confusion Matrix:')
print('              Predicted')
print('              0      1')
print(f'Actual 0:  {test_tn:4d}  {test_fp:4d}')
print(f'Actual 1:  {test_fn:4d}  {test_tp:4d}')
```

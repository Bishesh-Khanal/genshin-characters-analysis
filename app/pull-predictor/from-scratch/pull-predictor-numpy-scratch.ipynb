```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../../../data/cleaned_data/cleaned_pull_count.csv').drop("Unnamed: 0", axis=1)
```

```python
pd.set_option('display.max_rows', None)

pd.set_option('display.max_columns', None)

pd.set_option('display.width', None)

pd.set_option('display.max_colwidth', None)
```

```python
df.shape
```

```python
df_encoded = pd.get_dummies(df, columns=['3D-model', 'rarity', 'weapon', 'element', 'region', 'asc_stat_bonus_category'], prefix=['model', 'rarity', 'weapon', 'element', 'region', 'asc_bonus'], dtype=int)
```

```python
df_encoded.shape
```

```python
df_encoded.columns
```

```python
df_encoded.drop(['rarity_4', 'weapon_Sword', 'element_Pyro', 'region_Extra-Terresterial', 'asc_bonus_Pyro DMG Bonus', 'model_Tall Male', 'duplicate_rate', 'c6_rate', 'avg_copies_per_player'], axis = 1, inplace=True)
df_encoded['ones'] = 1
```

```python
df_encoded['pulled_count_TARGET_VARIABLE'] = np.log10(df_encoded['pulled_count_TARGET_VARIABLE']+1)

columns_to_normalize = ['lvl_90_HP', 'lvl_90_ATK', 'lvl_90_DEF', 'months_since_release', 'num_banners']
for col in columns_to_normalize:
    df_encoded[col] = (df_encoded[col] - df_encoded[col].min()) / (df_encoded[col].max() - df_encoded[col].min())
```

```python
df_encoded
```

```python
df_target = df_encoded['pulled_count_TARGET_VARIABLE']
df_inputs = df_encoded.drop('pulled_count_TARGET_VARIABLE', axis=1)
```

```python
print(df_target.shape)
print(df_inputs.shape)
```

```python
np.random.seed(42)
n_samples = len(df_inputs)
test_size = 0.3

indices = np.arange(n_samples)
np.random.shuffle(indices)

test_count = int(n_samples * test_size)
test_indices = indices[:test_count]
train_indices = indices[test_count:]

df_inputs_train = df_inputs.iloc[train_indices].reset_index(drop=True)
df_inputs_test = df_inputs.iloc[test_indices].reset_index(drop=True)
df_target_train = df_target.iloc[train_indices].reset_index(drop=True)
df_target_test = df_target.iloc[test_indices].reset_index(drop=True)

print(f'Training samples: {len(train_indices)}')
print(f'Test samples: {len(test_indices)}')
```

```python
np.random.seed(42)
values = np.random.randn(2, 51) * 0.01
biases = np.zeros(2)

result_train = pd.DataFrame({
    'Linear1': (df_inputs_train * values[0]).sum(axis=1) + biases[0],  # _train
    'Linear2': (df_inputs_train * values[1]).sum(axis=1) + biases[1]   # _train
})
result_train['ReLU1'] = result_train['Linear1'].apply(lambda x: max(0, x))
result_train['ReLU2'] = result_train['Linear2'].apply(lambda x: max(0, x))
result_train['Preds'] = result_train['ReLU1'] + result_train['ReLU2']
result_train['Loss'] = (result_train['Preds'] - df_target_train)**2  # _train
loss_last = result_train['Loss'].mean()

print(f'Initial loss: {loss_last}')
```

```python
result_train
```

```python
max_iterations = 3000
learning_rate = 0.01

for counter in range(max_iterations):
    grad_w1 = np.zeros(51)
    grad_w2 = np.zeros(51)
    grad_b1 = 0
    grad_b2 = 0
    
    for row in range(len(df_inputs_train)):
        error = result_train.loc[row, 'Preds'] - df_target_train.iloc[row]
        
        d_loss_d_pred = 2 * error
        
        # Gradient flows through the sum equally to both ReLU outputs
        d_pred_d_relu1 = 1
        d_pred_d_relu2 = 1
        
        d_relu1_d_linear1 = 1 if result_train.loc[row, 'Linear1'] > 0 else 0
        d_relu2_d_linear2 = 1 if result_train.loc[row, 'Linear2'] > 0 else 0
        
        grad_from_relu1 = d_loss_d_pred * d_pred_d_relu1 * d_relu1_d_linear1
        grad_w1 += grad_from_relu1 * df_inputs_train.iloc[row].values
        grad_b1 += grad_from_relu1
        
        grad_from_relu2 = d_loss_d_pred * d_pred_d_relu2 * d_relu2_d_linear2
        grad_w2 += grad_from_relu2 * df_inputs_train.iloc[row].values
        grad_b2 += grad_from_relu2
    
    grad_w1 /= len(df_inputs_train)
    grad_w2 /= len(df_inputs_train)
    grad_b1 /= len(df_inputs_train)
    grad_b2 /= len(df_inputs_train)
    
    values[0] -= learning_rate * grad_w1
    values[1] -= learning_rate * grad_w2
    biases[0] -= learning_rate * grad_b1
    biases[1] -= learning_rate * grad_b2
    
    result_train['Linear1'] = (df_inputs_train * values[0]).sum(axis=1) + biases[0]  # _train
    result_train['Linear2'] = (df_inputs_train * values[1]).sum(axis=1) + biases[1]  # _train
    result_train['ReLU1'] = result_train['Linear1'].apply(lambda x: max(0, x))
    result_train['ReLU2'] = result_train['Linear2'].apply(lambda x: max(0, x))
    result_train['Preds'] = result_train['ReLU1'] + result_train['ReLU2']
    result_train['Loss'] = (result_train['Preds'] - df_target_train)**2  # _train
    loss_new = result_train['Loss'].mean()
    
    if counter % 10 == 0:
        print(f'Iteration: {counter}, Loss: {loss_new:.6f}')
    
    if abs(loss_new - loss_last) < 1e-7:
        print(f'Converged at iteration {counter}')
        break
    
    loss_last = loss_new

print(f'Final loss: {loss_new}')
```

```python
result_test = pd.DataFrame({
    'Linear1': (df_inputs_test * values[0]).sum(axis=1) + biases[0],
    'Linear2': (df_inputs_test * values[1]).sum(axis=1) + biases[1]
})
result_test['ReLU1'] = result_test['Linear1'].apply(lambda x: max(0, x))
result_test['ReLU2'] = result_test['Linear2'].apply(lambda x: max(0, x))
result_test['Preds'] = result_test['ReLU1'] + result_test['ReLU2']
result_test['Loss'] = (result_test['Preds'] - df_target_test)**2
test_loss = result_test['Loss'].mean()

print(f'\n=== RESULTS ===')
print(f'Training Loss (MSE): {loss_new:.6f}')
print(f'Test Loss (MSE): {test_loss:.6f}')

train_rmse = np.sqrt(loss_new)
test_rmse = np.sqrt(test_loss)
print(f'\nTraining RMSE: {train_rmse:.6f}')
print(f'Test RMSE: {test_rmse:.6f}')

print('\nSample Test Predictions (log-space):')
print(result_test[['Preds']].head())
print('\nActual Test Values (log-space):')
print(df_target_test.head())

test_preds_original = 10**result_test['Preds'] - 1
test_actual_original = 10**df_target_test - 1
print('\nSample Test Predictions (original scale):')
for i in range(5):
    print(f'Predicted: {test_preds_original.iloc[i]:>10,.0f} | Actual: {test_actual_original.iloc[i]:>10,.0f}')
```
